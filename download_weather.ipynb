```python
import datetime as dt
from apcemm_forecasting.weather.download_weather import download_hrrr, convert_dir_to_nc

# HRRR Example:
# Every 6 hours, HRRR computes a forecast for the next 48h
# Here we take the forecast computed at 00:00, and want to know the data from 06:00 to 08:00

# Specify save location for HRRR .grib forecast data
path_hrrr_forecast = "/home/chinahg/GCresearch/contrailuncertainty/HRRR/"

# Specify save location for HRRR .nc forecast data (can be the same as above)
path_hrrr_forecast_nc = "/home/chinahg/GCresearch/contrailuncertainty/HRRR/"

# Specify the time at which the forecast was computed (UTC)
forecast_time = dt.datetime(2023, 8, 20, 0)

# Time at which we want the forecasted data for (UTC)
time2forecast_for = dt.datetime(2023, 8, 20, 6)

# Number of hours starting at time2forecast_for that we want data for
n_hours_download = 3

# Download the HRRR forecast data
download_hrrr(forecast_time, path_hrrr_forecast, time2forecast_for, n_hours=n_hours_download)

# Convert the forecast data to .nc files for all further uses
convert_dir_to_nc(path_hrrr_forecast + "20230820/", path_hrrr_forecast_nc, overwrite=False, is_HRRR=True)
```
