```python
# Imports
import numpy as np
import os as os
import netCDF4 as nc
import xarray as xr
import matplotlib.pyplot as plt
```

```python
# Open the file
filename = 'BOU-RS-01_2_RS92-GDP_002_20160414T190000_1-002-001.nc'
path2file = '/net/d15/data/vmeijer/GRUAN/BOU/2016/'+filename
nc_GRUAN = nc.Dataset(path2file, 'r', format='NETCDF4_CLASSIC')
```

```python
ds = xr.open_dataset(path2file)
```

```python
DAL746_alt = nc_GRUAN.variables['alt']
DAL746_RH = nc_GRUAN.variables['rh']

fig, ax = plt.subplots()

ax.plot(DAL746_RH[:], DAL746_alt[:])

ax.set_xlim((0,1))
ax.set_ylim((0, 13000))
ax.set_xlabel("RH")
ax.set_ylabel("Altitude [m]")

print(max(DAL746_RH[:]))
```
