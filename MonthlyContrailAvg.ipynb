```python
from netCDF4 import Dataset
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import numpy as np
%matplotlib inline
```

```python
f, ax = plt.subplots(1,1,figsize=(8,5),subplot_kw={'projection': ccrs.PlateCarree()})

with Dataset('/net/d14/data/seastham/contrail_masks_daily/coverage_v0_201911.nc','r') as nc:
    im = ax.pcolormesh(nc['lon'],nc['lat'],np.mean(nc['coverage'][...],axis=0),transform=ccrs.PlateCarree())
    ax.coastlines()
    
plt.colorbar(im,ax=ax)
im.set_clim(0,0.05)
```
