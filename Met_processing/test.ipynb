```python
import numpy as np
import xarray as xr
import os
import netCDF4 as nc
import tqdm
```

```python
pressures = ['200',
            '225', '250', '300',
            '350', '400', '450',
            '500', '550', '600',
            '650', '700', '750',
            '775', '800', '825',
            '850', '875', '900',
            '925', '950', '975',
            '1000']
pressures = np.array(pressures, dtype=float)
print(pressures)
```

```python

file_paths = []
for root, dirs, files in os.walk('/home/chinahg/GCresearch/ERA5_downloads'):
    for file in files:
        if file.endswith('.nc'):
            file_paths.append(os.path.join(root, file))
```

```python
file_paths = sorted(file_paths)
```

```python
for file in tqdm.tqdm(file_paths):
    ds = nc.Dataset(file,mode='r+')
    
    if 'pressure_level' in ds.variables.keys():
        str_ver = np.array(ds.variables['pressure_level'][:], dtype=str)
        if (str_ver == 'nan').any():
            ds.variables['pressure_level'][:] = pressures
        
    ds.close()
```
